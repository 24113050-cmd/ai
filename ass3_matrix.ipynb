```python
n = int(input("Enter the number of nodes: "))

adj_mat = []

for i in range(n):
    row = []
    for j in range(n):
        row.append(0)
    adj_mat.append(row)

e = int(input("Enter the number of edges: "))

for i in range(e):
    u = int(input("Enter the first node: "))
    v = int(input("Enter the second node: "))

    adj_mat[u-1][v-1] = 1
    adj_mat[v-1][u-1] = 1


h = []
print("Enter the Heurisstic values for the nodes: ")
for i in range(n):
    h.append(int(input("h(" + str(i + 1) + ") = ")))

start = int(input("Enter the starting node: "))
goal = int(input("Enter the goal node: "))

open_list = []
closed_list = []
parent = []

for i in range(n):
    parent.append(-1)

open_list.append(start)

while len(open_list) > 0:
    min_node = open_list[0]

    for node in open_list:
        if h[node - 1] < h[min_node - 1]:
            min_node = node
    open_list.remove(min_node)
    closed_list.append(min_node)

    print("Current node: ", min_node)
    print("open list: ", open_list)
    print("Closed list: ", closed_list)

    if min_node == goal:
        print("Goal node found!!")

        path = []
        current_node = goal

        while current_node != -1:
            path.append(current_node)
            current_node  = parent[current_node - 1]

        path.reverse()
        print("path: ", path)
        break

    for i in range(n):
        if adj_mat[min_node - 1][i] == 1:
            neighbour = i + 1
            if neighbour in closed_list:
                continue
            if neighbour not in open_list:
                parent[i] = min_node
                open_list.append(neighbour)
    print()
```
